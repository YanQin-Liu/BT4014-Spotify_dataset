```python
import matplotlib.pyplot as plt
import pandas as pd
```

```python
df = pd.read_csv("./Original csv/spotify_2015_2025_85k.csv")
df
```

```python
print((df['track_name'] == 'None').sum())
```

```python
# drop all the rows with "None" as track name
df = df[df['track_name'] != 'None'].copy()

# parse string to datetime
df['release_date'] = pd.to_datetime(df['release_date'])

# convert ms to sec and mins
df['duration_sec'] = df['duration_ms'] / 1000
df['duration_min'] = df['duration_ms'] / 60000

# convert mode and explicit to boolean
df['explicit'] = df['explicit'].astype(bool)
df['mode'] = df['mode'].map({0: 'minor', 1: 'major'})

# looks like minimum stream count is 1000, setting flag to take into account
df['low_stream_flag'] = df['stream_count'] == 1000

df
```

```python
df.to_csv('spotify_cleaned.csv', index=False)
```
