```python
import json
import csv

with open('playlist.json', 'r') as f:
    data = json.load(f)

results = []
for playlist in data['playlists']:
    pid = playlist['pid']
    for track in playlist['tracks']:
        results.append({
            'pid': pid,
            'artist_name': track['artist_name'],
            'track_name': track['track_name']
        })

with open('output.csv', 'w', newline='') as f:
    writer = csv.DictWriter(f, fieldnames=['pid', 'artist_name', 'track_name'])
    writer.writeheader()
    writer.writerows(results)

print(f"Done! {len(results)} rows written to output.csv")
```

```python
import csv

# Load unique tracks into a set of (track_name, artist_name) tuples
unique_tracks = set()
with open('unique_tracks.csv', 'r') as f:
    reader = csv.DictReader(f)
    for row in reader:
        unique_tracks.add((row['track_name'].strip(),
                          row['artist_name'].strip()))

# Filter output.csv keeping only tracks that exist in uniquetrack.csv
kept = []
removed = 0
with open('output.csv', 'r') as f:
    reader = csv.DictReader(f)
    for row in reader:
        key = (row['track_name'].strip(), row['artist_name'].strip())
        if key in unique_tracks:
            kept.append(row)
        else:
            removed += 1

# Write filtered results
with open('output_filtered.csv', 'w', newline='') as f:
    writer = csv.DictWriter(f, fieldnames=['pid', 'artist_name', 'track_name'])
    writer.writeheader()
    writer.writerows(kept)

print(
    f"Done! Kept {len(kept)} rows, removed {removed} rows → output_filtered.csv")
```
